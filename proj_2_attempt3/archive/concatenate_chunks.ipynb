```python
import json
import numpy as np

# Load data
with open('MAIN_DATA.json', 'r') as f:
    data = json.load(f)

# Merge chunks back into full papers
merged_papers = {}
char_lengths = []
chunk_counts = []

for paper_id, paper_data in data.items():
    # Join chunks with newline (preserves paragraph structure)
    full_text = '\n'.join(paper_data['chunks'])
    
    merged_papers[paper_id] = {
        'title': paper_data['name'],
        'text': full_text,
        'char_length': len(full_text),
        'chunk_count': len(paper_data['chunks'])
    }
    
    char_lengths.append(len(full_text))
    chunk_counts.append(len(paper_data['chunks']))

# Stats
print(f"Total papers: {len(merged_papers)}")
print(f"\nCharacter length stats:")
print(f"  Average: {np.mean(char_lengths):,.0f} chars")
print(f"  Median: {np.median(char_lengths):,.0f} chars")
print(f"  Min: {np.min(char_lengths):,.0f} chars")
print(f"  Max: {np.max(char_lengths):,.0f} chars")
print(f"\nEstimated tokens (chars/4):")
print(f"  Average: {np.mean(char_lengths)/4:,.0f} tokens")
print(f"  Max: {np.max(char_lengths)/4:,.0f} tokens")

# Save merged papers
with open('MERGED_PAPERS.json', 'w') as f:
    json.dump(merged_papers, f, indent=2)

print(f"\n✓ Saved to MERGED_PAPERS.json")
```

```python
import json

# Reload original data
with open('MAIN_DATA.json', 'r') as f:
    data = json.load(f)

# Merge papers with better error handling
merged_papers = {}

for paper_id, paper_data in data.items():
    try:
        # Join chunks
        full_text = '\n'.join(paper_data['chunks'])
        
        # Remove any problematic characters
        full_text = full_text.replace('\x00', '')  # Remove null bytes
        
        merged_papers[paper_id] = {
            'title': paper_data['name'],
            'text': full_text
        }
    except Exception as e:
        print(f"Error with paper {paper_id}: {e}")

# Save with ensure_ascii=True to avoid encoding issues
with open('MERGED_PAPERS_FIXED.json', 'w', encoding='utf-8') as f:
    json.dump(merged_papers, f, ensure_ascii=True, indent=2)

print(f"✓ Saved {len(merged_papers)} papers to MERGED_PAPERS_FIXED.json")
```
