```python
import pandas as pd

df = pd.read_csv("final_gold_standard.csv")

# Sample 10 Yes and 5 No
yes_sample = df[df["InGoldStandard"] == "Yes"].sample(n=10, random_state=40)
no_sample = df[df["InGoldStandard"] == "No"].sample(n=5, random_state=40)

final_15 = pd.concat([yes_sample, no_sample]).reset_index(drop=True)

# Print links
print("=== YOUR 15 PAPERS ===\n")
for _, row in final_15.iterrows():
    print(f"Title: {row['Title']}")
    print(f"Link: {row['Link (Use DOI or Title if missing)']}")
    print(f"InGoldStandard: {row['InGoldStandard']}")
    print()

# Save
final_15.to_csv("final_15_papers.csv", index=False)
print("Saved to final_15_papers.csv")
```
